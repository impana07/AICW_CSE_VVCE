```python
# accessing attributes and methods
class emp:
    name = 'Impana'
    salary = '25000'
    def show(self):
        print(self.name)
        print(self.salary)
```

```python
e1 = emp()
```

```python
# using get attribute - getattr instead of e1.name
print(getattr(e1,'name'))
```

```python
# has attribute - hasattr - return true if the object has attribute
print(hasattr(e1,'name'))
```

```python
print(hasattr(e1,'gender'))
```

```python
# set attribute - setattr - sets an attribute
setattr(e1, 'gender', 'F')
```

```python
print(getattr(e1, 'gender'))
```

```python
# delattr - delete attribute - deletes the attribute
delattr(e1,'gender')
```

```python
class Student:
    def __init__(self, usn, name, age):  # constructor
        #Instance variables
        self.usn = usn
        self.name = name
        self.age = age
    # instance method to access instance variable
    def show(self):
        print('USN', self.usn, 'Name', self.name, 'Age', self.age)
    # instance method to modify instance variable
    def update(self, usn, age):
        self.usn = usn
        self.age = age
```

```python
# create an object
stu = Student(40, 'Impana', 19)
stu.show()
```

```python
stu.update(17,20)
```

```python
stu.show()
```
