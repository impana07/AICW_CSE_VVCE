```python
from datetime import date
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    @classmethod
    def calculate_age(cls, name, birthyear):       # cls means classmethod
        return cls(name, date.today().year - birthyear)

    def show(self):
        print(self.name + "'s age is " + str(self.age))
```

```python
age = Student('Impana', 20)
age.show()
```

```python
by = Student.calculate_age('Impana', 2006)
by.show()
```

```python
# mini project
class BankAccount:
    def __init__(self, name, balance=0):
        #instance attributes
        self.name = name
        self.balance = balance
    # deposit money
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print("Deposit Successfuly!")
            print("Deposited:", amount)
        else:
            print("Invalid deposit amount")
    # withdraw money
    def withdraw(self, amount):
        if amount <= 0:
            print('Invalid withdrawl amount!')
        elif amount > self.balance:
            print("Insufficient blance!")
        else:
            self.balance -= amount
            print("Withdrawal successful!")
            print("Withdrawn:", amount)
    #check balance
    def check_balance(self):
        print("Current balance:", self.balance)
    # display account details
    def display_details(self):
        print('Account holder:', self.name)
        print("Balance: ", self.balance)
```

```python
acc1 = BankAccount('Impana', 1000)
```

```python
acc1.display_details()
```

```python
acc1.deposit(5000)
```

```python
acc1.check_balance()
```

```python
acc1.withdraw(3000)
```

```python
acc1.check_balance()
```

<h4>Inheritance, Polymorphism, Encapsulation</h4>

```python
# Inheritance
# Parent class
class Person:
    def person_info(self, name, age):
        print("Inside person class")
        print("Name:", name, 'Age:', age )
# Parent class 2
class Company:
    def company_info(self, company_name, location):
        print("Inside the company class")
        print('Company:', company_name,'Location:', location)
```

```python
# child class - inherits attributes from parent classes
class Employee(Person, Company):
    def Employee_info(self, salary, skill):
        print("Inside employee class - child class")
        print('Salary:', salary, 'Skill:', skill)
```

```python
# create an object of Employee
emp2 = Employee()
```

```python
# access the data
emp2.person_info('Impana', 20)
emp2.company_info('Edunet Foundation', 'Bengaluru')
emp2.Employee_info(25000, 'AIML Engineer')
```

```python
class Vehicle:
    def info(self):
        print('This is vehicle')
class Car(Vehicle):
    def car_info(self, name):
        print('Car name is:', name)
class Truck(Vehicle):
    def truck_info(self, name):
        print('Truck name is:', name)
```

```python
obj1 = Car()
obj1.info()
obj1.car_info('Tata')
```

```python
obj2 = Truck()
obj2.info()
obj2.truck_info('Eicher')
```

```python
# hybrid inheritance
class Vehicle:
    def vehicle_info(self):
        print("Inside vehicle class")
class Car(Vehicle):
    def car_info(self):
        print('Inside car class')
class Truck(Vehicle):
    def truck_info(self):
        print('Inside truck class')
class SportsCar(Car, Vehicle):
    def sports_car_info(self):
        print('Inside sports car class')
```

```python
# create an object
s_car = SportsCar()
s_car.vehicle_info()
s_car.car_info()
s_car.sports_car_info()
```

```python
s_car.truck_info()  # 'SportsCar' object has no attribute 'truck_info'
```

```python
# super() - function
class Company:
    def company_name(self):
        return 'Edunet Foundation'
class Employee(Company):
    def info(self):
        # call the super func
        c_name = super().company_name()
        print('She works at', c_name)
```

```python
emp3 = Employee()
emp3.info()
```

```python
# method overriding
class Vehicle:
    def max_speed(self):
        print('Max speed is 100 kmph')
class Car(Vehicle):
    # overriding the implementation of vehicle class - method in child class overrides the method of the parent class
    def max_speed(self):
        print('Max speed is 200 kmph')
```

```python
car = Car()
car.max_speed()
```

```python
# public member
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    # public instance method
    def show(self):
        # access public data members
        print('Name', self.name, 'Salary', self.salary)
```

```python
emp = Employee('Impana', 10000)
# access the public data members
print('Name', emp.name, 'Salary', emp.salary)
emp.show()
```

```python
# Private member
class Employee:
    def __init__(self, name, salary):
        self.name = name  # public data member
        self.__salary = salary  # private data member
```

```python
# create an object
emp = Employee('Impana', 5000)
# access the private data member
print('Salary', emp.__salary)
```

```python
# protected member
class Company:
    def __init__(self):
        # protected member
        self._project = 'NLP'
# child class
class Employee(Company):
    def __init__(self, name):
        self.name = name
        Company.__init__(self)
    def show(self):
        print("Employee Name:", self.name)
        # Access the protected member
        print('Working on project:', self._project)
```

```python
c = Employee('Impana')
c.show()
```

```python
c._project   # direct access
```

```python
class India():
    def capital(self):
        print('New Delhi is the capital city of India')
    def language(self):
        print('India has 22 official languages')
    def type(self):
        print('India is a developing nation!')
class USA():
    def capital(self):
        print('Capital of USA is Washington DC')
    def language(self):
        print('The official language in US is English')
    def type(self):
        print('USA is a developed nation!')
```

```python
ob_India = India()
ob_USA = USA()
for country in (ob_India, ob_USA):
    country.capital()
    country.language()
    country.type()
```

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    def move(self):
        print('Drive!')
class Boat:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    def move(self):
        print('Sails!')
class Plane:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    def move(self):
        print('Flies!')
```

```python
o1 = Car('Tata', 'Innova')
o1.move()
o2 = Boat('Sea Ray','Sundancer 320')
o2.move()
o3 = Plane('DRDO', 'Tejas')
o3.move()
```

```python
# Polymorphism with inheritance
class Bird:
    def intro(self):
        print('The are many types of birds!')
    def flight(self):
        print('Most of the birds can fly but some cannot!')
class Sparrow(Bird):
    def flight(self):
        print('Sparrows can fly!')
class Ostrich(Bird):
    def flight(self):
        print('Ostrich cannot fly!')
```

```python
ob_bird = Bird()
ob_bird.intro()
ob_bird.flight()
```

```python
ob_sparrow = Sparrow()
ob_sparrow.flight()
```

```python
ob_ostrich = Ostrich()
ob_ostrich.flight()
```

```python
# overloading 
class complex:
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def __add__(self, other):   # other is the complex number
        return self.a + other.a, self.b + other.b
ob1 = complex(1,2)
ob2 = complex(2,3)
ob3 = ob1 + ob2   # 1+2=3, 2+3=5
ob3  
```
